# 04 - Sentiment Analysis

Generate sentiment features and evaluate sentiment patterns across support tickets.

```python
import pandas as pd
pd.set_option("display.max_columns", 10)

df = pd.read_csv('D:\\customer_support-ai\\data\\customer_tickets_cleaned.csv')
df

import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
df["sentiment_score"] = df["clean_description"].apply(lambda x: analyzer.polarity_scores(x)["compound"])

df.to_csv(
    "customer_tickets_cleaned.csv",
    index=False,
    mode="w"
)


print("✅ Preprocessing Completed Successfully")
```
